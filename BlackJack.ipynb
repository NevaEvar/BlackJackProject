```python
import pandas as pd

blackJack_data = pd.read_csv("blackjack.csv")

blackJack_data
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X = blackJack_data.drop(columns=["winloss",
                                 "PlayerNo", 
                                 "blkjck",
                                 "plybustbeat", 
                                 "dlbustbeat", 
                                 "dlwinamt", 
                                 "ply2cardsum", 
                                 "card3", 
                                 "card4", 
                                 "card5", 
                                 "dealcard3",
                                 "dealcard4",
                                 "dealcard5",
                                 "sumofcards",
                                 "sumofdeal",
                                 "blkjck", 
                                 "plwinamt", 
                                 "dlwinamt",
                                 "Unnamed: 0"])
y = blackJack_data["winloss"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = DecisionTreeClassifier()
model.fit(X_train, y_train)
predictions = model.predict(X_test)

score = accuracy_score(y_test, predictions)
score

test = model.predict([[10, 10, 10, 9]])
test
```

```python
from sklearn.externals import joblib

model.fit(X, y)

joblib.dump(model, "blackjack-recomender_2.joblib")

```

```python
from sklearn.externals import joblib
from sklearn import tree

model = DecisionTreeClassifier()
model.fit(X, y)

tree.export_graphviz(model, out_file="blackjack-recomender_2.dot", 
                    feature_names=["card1", "card2", "dealcard1", "dealcard2", "jokin"], 
                    class_names=sorted(y.unique()), 
                    label="all", 
                    rounded=True, 
                    filled=True)
```
