```python
from tkinter import*
import random
from PIL import ImageTk, Image
import time
```

```python
#laskija pitää yllä kuinka monta korttia on otettu
counter = 0
#Laskija pitää yllä onko kortit luotu jo aikaisemmin
counterReset = 0

#tämä on väli aikainen korjaus 
scorePoints = 1000

dealerCounter = 0


global score
score = 0 

betCounter = 0

#luo pelaajan ja dealerin kortin piste määrä
playerPoints = 0 

dealerPoints = 0


    
#Luo peli ikkunan
gameRoot = Tk()
gameRoot.geometry("690x800")
gameRoot.config(bg="green")
    
#luo objectin, joka sisältää kuvan ja sen arvon
class CARD:
    
    def __init__(self, value, image, isAce):
        self.value = value
        self.image = image
        self.isAce = isAce

#lataa korit
cardNumberBack = ImageTk.Image.open("cards\pixel_cards\pixel_card_back.png")
cardNumber2 = ImageTk.Image.open("cards\pixel_cards\pixel_card_2.png")
cardNumber3 = ImageTk.Image.open("cards\pixel_cards\pixel_card_3.png")
cardNumber4 = ImageTk.Image.open("cards\pixel_cards\pixel_card_4.png")
cardNumber5 = ImageTk.Image.open("cards\pixel_cards\pixel_card_5.png")
cardNumber6 = ImageTk.Image.open("cards\pixel_cards\pixel_card_6.png")
cardNumber7 = ImageTk.Image.open("cards\pixel_cards\pixel_card_7.png")
cardNumber8 = ImageTk.Image.open("cards\pixel_cards\pixel_card_8.png")
cardNumber9 = ImageTk.Image.open("cards\pixel_cards\pixel_card_9.png")
cardNumber10 = ImageTk.Image.open("cards\pixel_cards\pixel_card_10.png")
cardNumberJack = ImageTk.Image.open("cards\pixel_cards\pixel_card_jack.png")
cardNumberQueen = ImageTk.Image.open("cards\pixel_cards\pixel_card_queen.png")
cardNumberKing = ImageTk.Image.open("cards\pixel_cards\pixel_card_king.png")
cardNumberAce = ImageTk.Image.open("cards\pixel_cards\pixel_card_ace.png")

#uudelleen muovaa kuvat
resizedBack = cardNumberBack.resize((100, 150), Image.ANTIALIAS)
resized2 = cardNumber2.resize((100, 150), Image.ANTIALIAS)
resized3 = cardNumber3.resize((100, 150), Image.ANTIALIAS)
resized4 = cardNumber4.resize((100, 150), Image.ANTIALIAS)
resized5 = cardNumber5.resize((100, 150), Image.ANTIALIAS)
resized6 = cardNumber6.resize((100, 150), Image.ANTIALIAS)
resized7 = cardNumber7.resize((100, 150), Image.ANTIALIAS)
resized8 = cardNumber8.resize((100, 150), Image.ANTIALIAS)
resized9 = cardNumber9.resize((100, 150), Image.ANTIALIAS)
resized10 = cardNumber10.resize((100, 150), Image.ANTIALIAS)
resizedJack = cardNumberJack.resize((100, 150), Image.ANTIALIAS)
resizedQueen = cardNumberQueen.resize((100, 150), Image.ANTIALIAS)
resizedKing = cardNumberKing.resize((100, 150), Image.ANTIALIAS)
resizedAce = cardNumberAce.resize((100, 150), Image.ANTIALIAS)

#luo uuden kokoiset kuvat
newBack = ImageTk.PhotoImage(resizedBack)
new2 = ImageTk.PhotoImage(resized2)
new3 = ImageTk.PhotoImage(resized3)
new4 = ImageTk.PhotoImage(resized4)
new5 = ImageTk.PhotoImage(resized5)
new6 = ImageTk.PhotoImage(resized6)
new7 = ImageTk.PhotoImage(resized7)
new8 = ImageTk.PhotoImage(resized8)
new9 = ImageTk.PhotoImage(resized9)
new10 = ImageTk.PhotoImage(resized10)
newJack = ImageTk.PhotoImage(resizedJack)
newQueen = ImageTk.PhotoImage(resizedQueen)
newKing = ImageTk.PhotoImage(resizedKing)
newAce = ImageTk.PhotoImage(resizedAce)

#lähettää kortit korttii konstruktoriin
global cardBack
cardBack = CARD(0, newBack, 0)
card2 = CARD(2, new2, 0)
card3 = CARD(3, new3, 0)
card4 = CARD(4, new4, 0)
card5 = CARD(5, new5, 0)
card6 = CARD(6, new6, 0)
card7 = CARD(7, new7, 0)
card8 = CARD(8, new8, 0)
card9 = CARD(9, new9, 0)
card10 = CARD(10, new10, 0)
cardJack = CARD(10, newJack, 0)
cardQueen = CARD(10, newQueen, 0)
cardKing = CARD(10, newKing, 0)
cardAce = CARD(11, newAce, 1)

    
#Laittaa pisteet esille ja score tekstin
currentScore = Label(gameRoot, text=scorePoints, bg="grey", font=("Helvetica", 18))
currentScore.place(x=500, y=50)
    
scoreText = Label(gameRoot, text="Score: ", bg="grey", font=("Helvetica", 18))
scoreText.place(x=410, y=50)
    
def CardReset():
        
    #Random choice command "randomCard.choice(cardList)"
    global randomCard
    global cardList
    randomCard = random.SystemRandom()
    #uusi ja lisätty kortin ottaja
    cardList = [card2, 
                card3,
                card4,
                card5,
                card6,
                card7,
                card8,
                card9,
                card10,
                cardJack,
                cardQueen,
                cardKing, 
                cardAce]
        
    def FullReset():
        #Resettaa/luo kortit/korttien paikat
        #Dealerin kortit
        #Tekee kaikki kortit globaleiksi
        global cardBack
        global dealerFirstCard, dealerSecondCard, dealerThirdCard, dealerFourthCard, dealerFifthCard, dealerSixthCard
        global playerFirstCard, playerSecondCard, playerThirdCard, playerFourthCard, playerFifthCard, playerSixthCard
        
        dealerFirstCard = Label(gameRoot, image=cardBack.image)
        dealerFirstCard.place(x=10, y=200)
    
        dealerSecondCard = Label(gameRoot, image=cardBack.image)
        dealerSecondCard.place(x=120, y=200)
    
        dealerThirdCard = Label(gameRoot, image=cardBack.image)
        dealerThirdCard.place(x=230, y=200)
    
        dealerFourthCard = Label(gameRoot, image=cardBack.image)
        dealerFourthCard.place(x=340, y=200)
    
        dealerFifthCard = Label(gameRoot, image=cardBack.image)
        dealerFifthCard.place(x=450, y=200)
    
        dealerSixthCard = Label(gameRoot, image=cardBack.image)
        dealerSixthCard.place(x=560, y=200)
        
        #Pelaajan kortit
        playerFirstCard = Label(gameRoot, image=cardBack.image)
        playerFirstCard.place(x=10, y=360)
    
        playerSecondCard = Label(gameRoot, image=cardBack.image)
        playerSecondCard.place(x=120, y=360)
    
        playerThirdCard = Label(gameRoot, image=cardBack.image)
        playerThirdCard.place(x=230, y=360)
    
        playerFourthCard = Label(gameRoot, image=cardBack.image)
        playerFourthCard.place(x=340, y=360)
    
        playerFifthCard = Label(gameRoot, image=cardBack.image)
        playerFifthCard.place(x=450, y=360)
    
        playerSixthCard = Label(gameRoot, image=cardBack.image)
        playerSixthCard.place(x=560, y=360)
        print(scorePoints)
        
        
    global ConfigReset    
    def ConfigReset():
        #resettaa nämä kortit default muotoon
        playerFirstCard.config(image=cardBack.image)
        playerSecondCard.config(image=cardBack.image)
        playerThirdCard.config(image=cardBack.image)
        playerFourthCard.config(image=cardBack.image)
        playerFifthCard.config(image=cardBack.image)
        playerSixthCard.config(image=cardBack.image)
            
        dealerFirstCard.config(image=cardBack.image)
        dealerSecondCard.config(image=cardBack.image)
        dealerThirdCard.config(image=cardBack.image)
        dealerFourthCard.config(image=cardBack.image)
        dealerFifthCard.config(image=cardBack.image)
        dealerSixthCard.config(image=cardBack.image)
        
        #nollaa kaiken
        global dealerCounter, counter, playerPoints, dealerPoints
        dealerCounter = 0
        counter = 0
        playerPoints = 0 
        dealerPoints = 0
        
            
    #tsekkaa pitääkö luoda uudet kortit vai resetata
    #aloita tästä peli looppi (ehkä)
    global counterReset
        
    if counterReset == 0:
        FullReset()
        counterReset = 1
    elif counterReset >> 0:
        ConfigReset()
    
def PlayerPointAdd(newPoint, isAce):
    global playerPoints
        
    playerPoints = playerPoints + newPoint
    print(playerPoints)
    if playerPoints <= 22 and isAce == 1:
        playerPoints = playerPoints - newPoint + 1
    elif playerPoints >= 22:
        print("bust")
        gameStatus.config(text="loss")
        global ConfigReset
        ConfigReset()
    else:
        print("fine")
                
    return

#laskee dealerin pisteet
def DealerPointAdd(newPoint, isAce):
    global dealerPoints
    dealerPoints = dealerPoints + newPoint
    if dealerPoints <= 22 and isAce == 1:
        dealerPoints = dealerPoints - newPoint + 1
    print(dealerPoints)
    return
    
    
#Antaa pelaajalle uuden kortin
def Hit():
    global counter
    counter = counter + 1
    newCard = randomCard.choice(cardList)
    if counter == 1:
        playerFirstCard.config(image=newCard.image)
    elif counter == 2:
        playerSecondCard.config(image=newCard.image)
    elif counter == 3: 
        playerThirdCard.config(image=newCard.image)
    elif counter == 4: 
        playerFourthCard.config(image=newCard.image)
    elif counter == 5:
        playerFifthCard.config(image=newCard.image)
    elif counter == 6:
        playerSixthCard.config(image=newCard.image)
        counter = 0
        
    PlayerPointAdd(newCard.value, newCard.isAce)
        
    return
    
#Antaa dealerille uuden kortin
def DealerHit():
    global dealerCounter
    dealerCounter = dealerCounter + 1
    newCard = randomCard.choice(cardList)
    if dealerCounter == 1:
        #luo myöhemmin käännettävän kortin
        global hiddenDealerCard
        hiddenDealerCard = newCard.image
    elif dealerCounter == 2:
        dealerSecondCard.config(image=newCard.image)
    elif dealerCounter == 3: 
        dealerThirdCard.config(image=newCard.image)
    elif dealerCounter == 4: 
        dealerFourthCard.config(image=newCard.image)
    elif dealerCounter == 5:
        dealerFifthCard.config(image=newCard.image)
    elif dealerCounter == 6:
        dealerSixthCard.config(image=newCard.image)
        counter = 0
        
    DealerPointAdd(newCard.value, newCard.isAce)
    return
    
        
    
#Ei anna pelaajalle uutta korttia
def Stand():
    global playerPoints, dealerPoints, hiddenDealerCard, scorePoints, currentBet
    
    #Näyttää piilotetun dealerin kortin
    dealerFirstCard.config(image=hiddenDealerCard)
    
    loop = True 
    while loop:
        if playerPoints > dealerPoints:
            print("add")
            DealerHit()
        elif dealerPoints > 21:
            gameStatus.config(text="win")
            scorePoints = scorePoints + currentBet * 2
            currentScore.config(text=scorePoints)
            loop = False
        elif playerPoints > dealerPoints:
            gameStatus.config(text="win")
            scorePoints = scorePoints + currentBet * 2
            currentScore.config(text=scorePoints)
            loop = False
        elif playerPoints == dealerPoints:
            gameStatus.config(text="draw")
            loop = False
        elif playerPoints < dealerPoints: 
            gameStatus.config(text="loss")
            currentBet = 0
            print(playerPoints)
            print(dealerPoints)
            loop = False
            
    return
    
#Tuplaa pelaajan potin
def Double():
    return
    
#Laittaa valitun määrän rahaa peliin ja laskee uuden scoren
def Bet():
    global currentBet, scorePoints
    currentBet = betSlider.get()
        
    #Luo/configaa entisen potin ja sen labelin
    def BetCreate():
            
        labelBet = Label(gameRoot, text="bet", font=("Helvetica", 20), bg="yellow")
        labelBet.place(x=315, y=550)
            
        global playedBet
        playedBet = Label(gameRoot, text=currentBet, font=("Helvetica", 20), bg="yellow")
        playedBet.place(x=365, y=550)
        
    def BetConfig():
        #resettaa betin ja kaikki kortit
        playedBet.config(text=currentBet)
        ConfigReset()
        
        #aloittaa uuden peli loopin
        DealerHit()
        DealerHit()
        
        
    #aloittaa uuden peli loopin
    DealerHit()
    DealerHit()
    
        
    #laskee uuden scoren
    scorePoints = scorePoints - currentBet
    currentScore.config(text=scorePoints)
            
    global betCounter
        
    if betCounter == 0:
        BetCreate()
        betCounter = 1
    elif betCounter >> 0:
        BetConfig()
        
        
    return

#luo ikkunan, josta saa tietoa onko voittamassa vai häviämässä
def Help():
    infoRoot = Tk()
    
    
    infoRoot.mainloop()
    
#luo pelin napit
hitButton = Button(gameRoot, text="hit", font=("Helvetica", 25), command=Hit)
hitButton.place(x=490, y=600)

standButton = Button(gameRoot, text="stand", font=("Helvetica", 25), command=Stand)
standButton.place(x=115, y=600)
    
doubleButton = Button(gameRoot, text="double", font=("Helvetica", 25), command=Double)
doubleButton.place(x=290, y=600)
    
resetButton = Button(gameRoot, text="reset/start",font=("Helvetica", 25), command=CardReset)
resetButton.place(x=265, y=700)
    
betButton = Button(gameRoot, text="bet", font=("Helvetica", 20), bg="grey", command=Bet)
betButton.place(x=490, y=700)
    
betSlider = Scale(gameRoot, from_=0, to=scorePoints, orient=HORIZONTAL, bg="grey")
betSlider.place(x=115, y=700)

gameStatus = Label(gameRoot, text="X", font=("Helvetica", 25))
gameStatus.place(x=290, y=100)

gameInfo = Button(gameRoot, text="info/help)", font=("Helvetica", 20), command=Help)
gameInfo.place(x=20, y=20)
    
gameRoot.mainloop()
```
